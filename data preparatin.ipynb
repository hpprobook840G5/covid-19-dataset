```python
temp1=df[df["Province_State"].isna()]
temp2=df[df["Province_State"].notna()]
```

```python
temp1_date=temp1.groupby("Date").agg(total_confirmed=("Confirmed","sum"),
                                                total_active=("Active","sum"),
                                                total_recovered=("Recovered","sum"),
                                                total_deaths=("Deaths","sum"))
temp2_date=temp2.groupby("Date").agg(total_confirmed=("Confirmed","sum"),
                                                total_active=("Active","sum"),
                                                total_recovered=("Recovered","sum"),
                                                total_deaths=("Deaths","sum"))
```

```python
date_df=pd.concat([temp1_date,temp2_date])
date_df=date_df.reset_index()
daily_df=date_df.groupby("Date").agg(total_confirmed=("total_confirmed","sum"),
                                                total_active=("total_active","sum"),
                                                total_recovered=("total_recovered","sum"),
                                                total_deaths=("total_deaths","sum"))

```

```python
import matplotlib.pyplot as plt
x=daily_df.index
y=daily_df["total_confirmed"]

plt.plot(x,y,color="red",marker="o",linestyle="-",linewidth=0.2)
plt.xlabel("Month")
plt.ylabel("Number of Confirmed Cases")
plt.title("Number of Confirmed Cases over time")
plt.show()
```

```python
x=daily_df.index
y=daily_df["total_deaths"]
plt.plot(x,y,color="red")
plt.xlabel("Months")
plt.ylabel("Number of deaths")
plt.title("Number of deaths over time")
plt.show()
```

```python
x=daily_df.index
y=daily_df["total_recovered"]
plt.plot(x,y,color="red")
plt.xlabel("Month")
plt.ylabel("Number of Recoveries")
plt.title("Change in Recovery over time")
plt.show()
```
